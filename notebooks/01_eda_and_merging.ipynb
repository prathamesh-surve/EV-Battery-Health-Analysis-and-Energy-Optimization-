```python
import pandas as pd
import glob
import os

# 1. Locate all CSV files in data/raw
csv_files = glob.glob("../data/raw/*.csv")
print(f"Found {len(csv_files)} raw battery data files to combine.")

# 2. Read each individual file and append it to a temporary list
df_list = []
for file in csv_files:
    df = pd.read_csv(file)
    df_list.append(df)

# 3. Concatenate all rows into one master dataframe
combined_df = pd.concat(df_list, ignore_index=True)

# 4. Convert 'start_time' using format='mixed' to cleanly parse arbitrary formats
if 'start_time' in combined_df.columns:
    combined_df['start_time'] = pd.to_datetime(combined_df['start_time'], format='mixed')
    combined_df = combined_df.sort_values(by='start_time').reset_index(drop=True)

print(f"✅ Successfully merged dataset! Total combined rows and columns: {combined_df.shape}")

# 5. Export a master copy back to your raw data storage folder
combined_df.to_csv("../data/raw/unified_raw_battery_data.csv", index=False)
print("Saved master file as 'unified_raw_battery_data.csv'")
combined_df.head()
```
